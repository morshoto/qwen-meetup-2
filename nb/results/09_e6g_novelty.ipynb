Does a reminder to produce a different program increase novelty?

These notebooks use **curated aggregate data**, transcribed from the cited reports. They require no model, network, GPU, Docker, or local experiment artifacts. Counts describe the recorded runs through November 5, 2026. Run all cells from a fresh kernel.

A **verified Oracle** is a model-authored program that passes the synthetic-task checks. A **strict new Oracle** also passes the registered provenance and novelty rules. Neither is evidence that the model independently solved a real ARC task.

```python
import matplotlib.pyplot as plt
import pandas as pd
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams.update({"figure.figsize": (8, 4), "figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False})
```

## How the experiment worked

G0 used the existing generation and repair loop. G1 additionally received a domain-neutral duplicate reminder. Each arm used a separate copy of the same initial pool.

## Results

Source: [E6-G report](../../docs/e6g-autonomous-curriculum-diversity.md).

The data below are aggregate observations, not new model runs.

```python
data = pd.DataFrame({'Arm': ['G0 baseline', 'G1 feedback'], 'Completed': [2, 1], 'Valid': [0, 1], 'Novel': [0, 0]})
ax = data.set_index('Arm')[['Completed', 'Valid', 'Novel']].plot.bar(rot=0, title='Live smoke results')
ax.set_ylabel('Candidates')
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
data
```

## What to take away

Both arms produced zero verified novel Oracles. G1 exhausted its sandbox cap before the second candidate, so the arms have unequal completed counts. This tiny smoke found no demonstrated gain and cannot establish that feedback has no effect.
