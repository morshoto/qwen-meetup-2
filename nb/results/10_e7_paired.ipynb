```python
data = pd.DataFrame({'Arm': ['G0 baseline', 'G1 feedback'], 'Completed': [2, 2], 'Novel': [0, 0], 'AST duplicates': [0, 2]})
ax = data.set_index('Arm')[['Completed', 'Novel', 'AST duplicates']].plot.bar(rot=0, title='Two paired slots per arm')
ax.set_ylabel('Candidates')
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
data
```

## What to take away

Neither arm produced a novel Oracle. G1 repeated existing source/AST in both slots; its two evaluated candidates also matched sampled behavior in the independent audit. G0 had no behavior fingerprints, so its behavior duplicate rate was not measured. Two pairs do not establish a causal effect.
