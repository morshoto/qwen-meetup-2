```python
data = pd.DataFrame({'Outcome': ['Attempted tasks', 'Verified tasks', 'Target verified tasks'], 'Count': [19, 0, 5]})
ax = data.set_index('Outcome')[['Count']].plot.bar(rot=0, title='Collection stopped at the budget limit')
ax.set_ylabel('Tasks')
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
data
```

## What to take away

The collection attempted 19 tasks and accepted zero verified solutions before its call or reserved-token budget was exhausted. Sixteen selected candidates failed independent verification, two tasks had no selected candidate, and the last stopped before dispatch. Training did not run.
